```python
import pandas as pd
```

```python
all_data=pd.read_csv(r"C:\Users\shoba\stock_analysis\data\raw\all_stocks_raw.csv")

all_df=pd.DataFrame(all_data)

all_df.head()
```

```python
#Dropping the unnamed column

all_df.drop(columns=["Unnamed: 0"], inplace=True)
all_df.head()
```

```python
# Correcting tha columns order
col_ord=["Date", "Ticker", "Company", "Sector", "Open", "High", "Low", "Close", "Volume"]
all_df=all_df[col_ord]
all_df.head()
```

```python
#Correcting the date format
all_df["Date"] = pd.to_datetime(all_df["Date"], errors="coerce")
all_df["Volume"]=all_df["Volume"].astype("Int64")
all_df.dtypes
```

```python
# Sorting the data by date for each ticker

all_df.sort_values(by=['Ticker','Date'],inplace=True)
all_df.head(20)
```

```python
# Chhecking Null Values
all_df.isnull().sum()
```

```python
price_cols=["Open", "High", "Low", "Close"]
all_df[price_cols]=all_df[price_cols].ffill()
all_df["Volume"]=all_df["Volume"].fillna(0)
all_df.dropna(subset=['Date'],axis=0,inplace=True)
```

```python
all_df.isnull().sum()
```

```python
#Check For Duplicates

all_df.duplicated().sum()
```

```python
# Creating more columns for analysis

all_df['daily_return']=all_df.groupby('Ticker')['Close'].pct_change()
all_df['year']=all_df['Date'].dt.year
all_df['month']=all_df['Date'].dt.month
all_df["day_of_week"]=all_df['Date'].dt.dayofweek
all_df.head()
```
