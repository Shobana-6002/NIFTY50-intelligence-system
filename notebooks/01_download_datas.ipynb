```python
start_date=datetime.today().replace(year=datetime.today().year-5).strftime('%Y-%m-%d')
end_date=datetime.today().strftime('%Y-%m-%d')
```

```python
for t in ticker:
    print(f"Downloading data for {t}...")
    try:
        data = yfinance.download(t, start=start_date, end=end_date)
        data.columns = data.columns.get_level_values(0)
        if data.empty:
            print(f"No data found for {t}. Skipping.")
            continue
        if not data.empty:
            data.reset_index(inplace=True)
            data['Ticker'] = t
            data['Company'] = stock_info[t]['company']
            data['Sector'] = stock_info[t]['sector']
       
        filename = stock_info[t]['company'].replace(" ", "_") + ".csv"
        
        data.to_csv(os.path.join(r"C:\\Users\\shoba\\stock_analysis\\data\\raw", filename), index=False)
    except Exception as e:
        print(f"Error downloading data for {t}: {e}")
        continue
```

```python
nifty_ticker = "^NSEI"
print(f"Downloading data for {nifty_ticker}...")
nifty_data = yfinance.download(nifty_ticker, start=start_date, end=end_date)
nifty_data.reset_index(inplace=True)
nifty_data.to_csv(os.path.join(r"C:\\Users\\shoba\\stock_analysis\\data\\raw", "Nifty_50.csv"), index=False)
```

```python
all_dataframes = []
files=os.listdir(r"C:\\Users\\shoba\\stock_analysis\\data\\raw")
for file in files:
    if file == "Nifty_50.csv":
        continue
    if file == "all_stocks_raw.csv":
        continue
    if file.endswith(".csv"):
        full_path = os.path.join(r"C:\\Users\\shoba\\stock_analysis\\data\\raw", file)
        df = pd.read_csv(full_path)
        all_dataframes.append(df)
        print(f"Loaded {file} with {len(df)} rows.")

master_df = pd.concat(all_dataframes)

save_path=os.path.join(r"C:\\Users\\shoba\\stock_analysis\\data\\raw", "all_stocks_raw.csv")
master_df.to_csv(save_path)

print("\nMaster file created successfully!")
print("Total rows:", master_df.shape[0])
print("Total columns:", master_df.shape[1])
```
